```python
def generateMatrix(n: int) -> list[list[int]]:
        matrix = [[0]*n for _ in range(n)]
        res = []
        count = 0
        rt_wall = n
        bt_wall = n
        lt_wall = -1
        up_wall = 0
        RIGHT, DOWN, LEFT, UP = 0, 1, 2, 3
        DIRECTION = RIGHT
        i,j=0,0
        
        while len(res)!=n*n:
            if DIRECTION==RIGHT and len(res)!=n*n:
                while j<rt_wall:
                    count+=1
                    matrix[i][j] = count
                    res.append(count)
                    j+=1
                rt_wall-=1
                j=rt_wall
                i+=1
                DIRECTION = DOWN
            print(matrix)    
            
            if DIRECTION==DOWN and len(res)!=n*n:
                while i<bt_wall:
                      count+=1
                      matrix[i][j] = count
                      res.append(count)
                      i+=1
                bt_wall-=1
                i=bt_wall
                j-=1
                DIRECTION = LEFT

            if DIRECTION==LEFT and len(res)!=n*n:
                while j>lt_wall:
                    count+=1
                    matrix[i][j] = count
                    res.append(count)
                    j-=1

                lt_wall+=1
                j=lt_wall
                i-=1
                DIRECTION=UP
            
            if  DIRECTION==UP and len(res)!=n*n:
                while i>up_wall:
                    count+=1
                    matrix[i][j] = count
                    res.append(count)
                    i-=1
                up_wall+=1
                i=up_wall
                j+=1
                DIRECTION=RIGHT


        return matrix        
```

```python
def countPrimes(n: int) -> int:
        def isPrime(n):
            if n <=1:
               return False
            for i in range(2,n):
                if n%i==0:
                   return False
            return True

        count = 0
        if n<=1:
           return 0
        for i in range(2,n):
            if isPrime(i):
               count+=1
        return count 
```

```python

from typing import List


def minJumps(nums: List[int]) -> int:
    def isPrime(n):
        if n<=1:
            return False
        for i in range(2,n):
            if n%i==0:
                return False
        return True
    steps = 0
    for i in range(len(nums)):
        if isPrime(nums[i]):
            print('nums[i] is prime', nums[i], 'at index', i)
            for j in range(len(nums)-1, i, -1):
                if nums[j] % nums[i] == 0:
                    print('nums[j] is divisible by nums[i]', nums[j], 'at index', j)
                    i = j + 1
                    steps+=1
                    break
            
    return steps
 
```
